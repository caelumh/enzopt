### Loading the EnzymeML document for PySCeS fitting

The EnzymeML document from the previous Notebook can be used directly with PySCeS for fitting the kinetic parameters of the enzymatic cascade reaction directly from PyEnzyme using the implemented thin layer interface.

#### EnzymeML document

```python
enzmldoc = pe.read_enzymeml('./data/output/pgm-eno.json')
```

```python
enzmldoc
```

```python
enzmldoc.parameters
```

```python
pe.summary(enzmldoc)
```

#### PySCeS fitting

After instantiating the PySCeS thin layer, multiple fitting algorithms that can be passed with the `method` argument to the `optimize` method are available for parameter estimation.

```python
tl_psc = ThinLayerPysces(enzmldoc=enzmldoc)
```

```python
res = tl_psc.optimize()
```

```python
res.params
```

```python
# other optimizer method
res2 = tl_psc.optimize(method='nelder')
```

```python
res2.params
```

The results of the fitting can be visualized using the `plot` method of PyEnzyme.

```python
f, ax = pe.plot(
    enzmldoc,
    thinlayer=tl_psc,
    show=True,
 )
```

### Identifyability analysis

If desired, using the Python package `identifiability`, the confidence intervals can be calculated and plotted on the basis of a profile-likelihood.  
Based on `lmfit` (including custom functions to select the range for parameter scanning and for plotting the profile likelihood), the chi-squared distribution is used to assess the significance.

```python
from identifiability import ConfidenceInterval
```

```python
ci = ConfidenceInterval(tl_psc.minimizer, res)
```

```python
# takes a while
ci.calc_all_ci(mp=False)
```

```python
ci.plot_all_ci()
```
